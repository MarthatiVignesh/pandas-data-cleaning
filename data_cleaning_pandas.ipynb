```python
df = pd.read_csv("basic-data (2).csv", skiprows=1)
```

```python
df.columns
```

```python
df["Age"].value_counts().sort_index()
```

```python
df[df["Age"] == 0]
```

```python
(df["Age"] == 0).sum()
```

```python
median_age = df.loc[df["Age"] > 0, "Age"].median()

print(median_age)
```

```python
df.loc[df["Age"] == 0, "Age"] = median_age
```

```python
(df["Age"] == 0).sum()
```

```python
df["Email"].isna().sum()
```

```python
df["Email"] = df["Email"].fillna("test@example.com")
```

```python
df["Email"].isna().sum()
```

```python
df["Name"].isna().sum()
```

```python
df["Name"] = df["Name"].fillna("default")
```

```python
df["Name"].isna().sum()
```

```python
df.isna().sum()
```

```python
df.to_csv("cleaned_basic_data.csv", index=False)
```
